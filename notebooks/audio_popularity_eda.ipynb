```python
numerical_features = [
    "Duration_Min",
    "Danceability",
    "Energy",
    "Valence",
    "Acousticness",
    "Instrumentalness",
    "Speechiness",
    "Tempo_BPM",
    "Loudness_dB",
    "Popularity"
]

for column in numerical_features:
    plt.figure(figsize=(8, 4))
    
    sns.boxplot(data=df, x=column)
    
    plt.title(f"Outlier Analysis - {column}")
    plt.xlabel(column)
    
    plt.show()
```

```python
outlier_summary = []

for column in numerical_features:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    
    outlier_summary.append({
        "Feature": column,
        "Outlier_Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

### Observation

Outlier analysis identified unusual observations across several audio features. These values were retained because they may represent genuine differences between songs rather than data-entry errors.

```python
correlation_matrix = df[numerical_features].corr()

correlation_matrix
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Audio Features and Popularity")
plt.show()
```

### Observation

The correlation heatmap shows the relationships between the numerical audio features and song popularity. Some features show positive or negative associations with popularity, while others have relatively weak linear relationships.

```python
df["Popularity_Category"] = pd.cut(
    df["Popularity"],
    bins=[-1, 39, 69, 100],
    labels=["Low", "Medium", "High"]
)
```

```python
df[["Track_Name", "Popularity", "Popularity_Category"]].head(10)
```

```python
df["Popularity_Category"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Popularity_Category",
    order=["Low", "Medium", "High"]
)

plt.title("Distribution of Songs by Popularity Category")
plt.xlabel("Popularity Category")
plt.ylabel("Number of Songs")
plt.show()
```

### Observation

Songs were categorized into Low, Medium, and High popularity groups to make popularity easier to compare. The distribution shows how the songs are spread across different popularity levels.

```python
popularity_features = [
    "Danceability",
    "Energy",
    "Valence",
    "Acousticness",
    "Instrumentalness",
    "Speechiness",
    "Tempo_BPM",
    "Loudness_dB"
]

category_means = (
    df.groupby("Popularity_Category", observed=False)[popularity_features]
      .mean()
      .round(2)
)

category_means
```

```python
category_means.T.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Average Audio Features by Popularity Category")
plt.xlabel("Audio Feature")
plt.ylabel("Average Value")
plt.xticks(rotation=45)
plt.legend(title="Popularity")
plt.show()
```

### Observation

The comparison of popularity categories shows differences in the average audio characteristics of Low, Medium, and High popularity songs. Some features show clearer variation across popularity groups, while others remain relatively similar.

```python
top_10_songs = (
    df.sort_values(by="Popularity", ascending=False)
      [["Track_Name", "Artist", "Genre", "Release_Year", "Popularity"]]
      .head(10)
)

top_10_songs
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_songs,
    x="Popularity",
    y="Track_Name"
)

plt.title("Top 10 Most Popular Songs")
plt.xlabel("Popularity Score")
plt.ylabel("Song")
plt.show()
```

### Observation

The Top 10 analysis identifies the highest-popularity songs in the dataset. The results also show which artists and genres are represented among the most popular tracks.

```python
artist_popularity = (
    df.groupby("Artist")
      .agg(
          Average_Popularity=("Popularity", "mean"),
          Song_Count=("Track_ID", "count")
      )
      .sort_values("Average_Popularity", ascending=False)
)

artist_popularity
```

```python
top_10_artists = artist_popularity.head(10)

top_10_artists
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_10_artists.reset_index(),
    x="Average_Popularity",
    y="Artist"
)

plt.title("Top 10 Artists by Average Song Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Artist")
plt.show()
```

### Observation

The artist-level analysis shows differences in average song popularity across artists. Comparing average popularity with song count helps identify artists who perform consistently well rather than simply having a larger number of songs.
